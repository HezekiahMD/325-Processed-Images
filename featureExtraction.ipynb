# Feature Extraction

```python
#Extracts features from 224 x 224 image
#returns array of avg R,G,B values in an 8x8 area
def extractFeatures(image:ImageFile):

    avgRedVals = []
    avgGreenVals = []
    avgBlueVals = []
    features = []

    for x in range(0,224,8):
        for y in range(0,224,8):

            avgRedVal = 0
            avgGreenVal = 0
            avgBlueVal = 0

            for i in range(0,8):
                for j in range(0,8):

                    r, g, b = image.getpixel((x+i, y+j))

                    avgRedVal += r
                    avgGreenVal += g
                    avgBlueVal += b

            avgRedVal = avgRedVal / 64
            avgGreenVal = avgGreenVal / 64
            avgBlueVal = avgBlueVal / 64

            avgRedVals.append(avgRedVal)
            avgGreenVals.append(avgGreenVal)
            avgBlueVals.append(avgBlueVal)

    features = avgRedVals + avgGreenVals + avgBlueVals
    return features
```
